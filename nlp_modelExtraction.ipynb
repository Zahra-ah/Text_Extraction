```python
%%writefile text_extraction.py
import streamlit as st
import spacy
#extract within text ----->highlight
from spacy import displacy
st.title("text Extraction application")
st.write("please,enter the text...")
text=st.text_area("enter the text here....",height=150)
nlp_model=spacy.load("en_core_web_lg")
#define function for text extraction
def text_extraction(text):
  document=nlp_model(text)
  entities=[(ent.text,ent.label_) for ent in document.ents]
  return entities
def show_entity(document):
  json_file=displacy.render(document,style="ent",jupyter=False)
  return json_file
  st.markdown(json_file,unsafe_allow_html=True)
if st.button("extract"):
  if text :
    document=nlp_model(text)
    entities=text_extraction(text)
    st.write(entities)
    st.subheader("highlight")
    st.markdown(show_entity(document),unsafe_allow_html=True)
```
